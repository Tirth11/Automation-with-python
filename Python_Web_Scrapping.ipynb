```python
import time
import random
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options

# Setting up Chrome options to avoid manually downloading the Chrome driver
options = Options()
options.add_argument("--headless")  # Runs Chrome in headless mode
options.add_argument("--disable-gpu")  # Disable GPU acceleration
options.add_argument("--no-sandbox")  # Disable sandboxing for better performance
options.add_argument("start-maximized")  # Maximizes the window on startup

# Create a service to automatically fetch the required driver for Chrome
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# Define the login function
def login_to_amazon(username, password):
    driver.get("https://www.amazon.in/ap/signin")

    # Locate the username field and enter the credentials
    driver.find_element(By.ID, "ap_email").send_keys(username)
    driver.find_element(By.ID, "continue").click()
    time.sleep(random.uniform(2, 5))

    driver.find_element(By.ID, "ap_password").send_keys(password)
    driver.find_element(By.ID, "signInSubmit").click()
    time.sleep(random.uniform(3, 6))

# Function to scrape a category page
def scrape_category(url, category_name):
    driver.get(url)
    time.sleep(random.uniform(3, 6))

    products = []

    # Loop through multiple pages of the category
    for page in range(1, 6):  # You can adjust this based on how many pages you want to scrape
        print(f"Scraping page {page} in {category_name}")

        # Wait for products to load
        time.sleep(random.uniform(5, 10))

        # Get product data from the current page
        product_elements = driver.find_elements(By.CSS_SELECTOR, ".zg-item-immersion")

        for product in product_elements:
            try:
                product_name = product.find_element(By.CSS_SELECTOR, ".p13n-sc-truncated").text
                product_price = product.find_element(By.CSS_SELECTOR, ".p-price .a-offscreen").text
                sale_discount = product.find_element(By.CSS_SELECTOR, ".a-size-small .a-color-price").text
                rating = product.find_element(By.CSS_SELECTOR, ".a-icon-alt").text
                best_seller_rating = product.find_element(By.CSS_SELECTOR, ".zg-badge-text").text
                sold_by = product.find_element(By.CSS_SELECTOR, ".a-color-secondary .a-size-base").text
                ship_from = product.find_element(By.CSS_SELECTOR, ".a-size-small.a-color-secondary").text

                # If any fields are missing, continue to the next product
                if not product_name or not product_price:
                    continue

                products.append({
                    'Product Name': product_name,
                    'Product Price': product_price,
                    'Sale Discount': sale_discount,
                    'Best Seller Rating': best_seller_rating,
                    'Ship From': ship_from,
                    'Sold By': sold_by,
                    'Rating': rating,
                    'Category Name': category_name
                })

            except Exception as e:
                print(f"Error while scraping a product: {e}")
                continue

        # Navigate to the next page
        try:
            next_button = driver.find_element(By.CSS_SELECTOR, ".a-last a")
            next_button.click()
            time.sleep(random.uniform(5, 10))
        except Exception as e:
            print(f"No more pages available or error navigating: {e}")
            break

    return products

# Main scraping function
def scrape_amazon_best_sellers():
    login_to_amazon('your_email@example.com', 'your_password')

    category_urls = [
        "https://www.amazon.in/gp/bestsellers/kitchen/ref=zg_bs_nav_kitchen_0",
        "https://www.amazon.in/gp/bestsellers/shoes/ref=zg_bs_nav_shoes_0",
        "https://www.amazon.in/gp/bestsellers/computers/ref=zg_bs_nav_computers_0",
        "https://www.amazon.in/gp/bestsellers/electronics/ref=zg_bs_nav_electronics_0",
        # Add more categories as needed
    ]

    all_products = []

    for category_url in category_urls:
        category_name = category_url.split('/')[-2]  # Extract category name from the URL
        category_products = scrape_category(category_url, category_name)
        all_products.extend(category_products)

    # Convert data to DataFram
```
